```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/PMID_Publisher.csv')

# Group the data by publisher and count the number of articles using reset_index 
publisher_counts = df.groupby('publisher')['PMID_issn'].count().reset_index(name='count')
# Get the journal with the highest number of titles

top_publisher = publisher_counts.nlargest(1, 'count')

print(top_publisher)
```

```python
print(publisher_counts)
```

```python
#top 10 journals
top_10_publisher = publisher_counts.nlargest(10, 'count')

print(top_10_publisher)
```

```python
# Create a chart
plt.figure(figsize=(10, 6))
plt.barh(top_10_publisher['publisher'], top_10_publisher['count'])
plt.xlabel('Number of Papers')
plt.ylabel('Publisher')
plt.title('Top 10 Publishers by Number of Titles')
plt.show()
```
