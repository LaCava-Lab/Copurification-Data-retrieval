```python
import matplotlib.pyplot as plt
import plotly.graph_objects as go

# make a df of successfull from first try
nan_dropped = nan.dropna(subset=['publisher'])
first_try = len(nan_dropped)
# Filter rows where the publisher is missing
no_publisher = nan[nan['publisher'].isnull()]

# Further filter those where the DOI is also missing
no_doi = no_publisher[no_publisher['doi'].isnull()]

# Counts for Sankey plot
total_fetched = len(nan)
publisher_ok = len(fin)
non_retrieved = len(final_missing)
rows_without_publisher = len(no_publisher)
rows_without_doi = len(no_doi)




# Define nodes and their labels
node_labels = ["error",'no doi', '1st time successful', 'total fetched', 'final retrieved', 'Failed to retrieve']

# Define the nodes and links
nodes = dict(
    label = node_labels,
    color = ["green", "blue", "orange", 'black', 'green', 'black'],
    pad = 15,
    thickness = 15,
    line = dict(color = "black", width = 0.5)
)

links = dict(
    source = [0, 1, 2, None, 3, 3],  # Indices in `node_labels`
    target = [3, 3, 3, None, 5, 4],
    value = [rows_without_publisher - rows_without_doi, rows_without_doi, first_try, total_fetched, non_retrieved, publisher_ok]
)

# Create the Sankey plot
fig = go.Figure(data=[go.Sankey(
    node=nodes,
    link=links
)])

# Update the layout
fig.update_layout(title_text="Publisher availability", font_size=20)
fig.show()
```
