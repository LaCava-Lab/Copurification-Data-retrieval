```python
import logging
import requests
import json
import time
import pandas as pd
from habanero import Crossref
cr = Crossref()
from functools import partial
from tenacity import retry, stop_after_delay, wait_fixed, retry_if_exception_type
import os
import matplotlib.pyplot as plt
```

```python

os.chdir('C:/Users/mkha1/Desktop/GQP/fnl_Scripts/PMID_lists/X-rey_chryst')
df = pd.read_csv('FINAL.csv')
total_rows = len(df)
rows_with_na = df['publisher'].isnull().sum()  # Rows with any NA values
rows_without_na = total_rows - rows_with_na  # Rows without any NA values
```

```python
# Data to plot
labels = ['Missing', 'Success']
sizes = [rows_with_na, rows_without_na]
colors = ['lightcoral', 'lightskyblue']
explode = (0.1, 0)  # explode the 1st slice

# Plot
plt.figure(figsize=(8,6))
patches, texts, autotexts = plt.pie(sizes, explode=explode, labels=labels, colors=colors,
autopct='%1.1f%%', shadow=True, startangle=140)

plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.title('pipe result')

# Customize legend
legend_labels = [f'{label}: {size} ({perc})' for label, size, perc in zip(labels, sizes, [f"{(s/total_rows)*100:.1f}%" for s in sizes])]
plt.legend(patches, legend_labels, title="Row counts", loc="best")

plt.show()
```
