```python
import matplotlib.pyplot as plt
import seaborn as sns

# numbers
data = {'DOI OK': 82101, 'not OK': 2314}

# Create pie chart with total
plt.pie(data.values(), labels=data.keys(), autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))

# Calculate total
total = sum(data.values())

# Add legend with total label
plt.title('Xreff yield Using DOI ids in habanero')
plt.legend(title='Total: {}'.format(total))

# Set aspect ratio to be equal so that pie is drawn as a circle
plt.axis('equal')

# Apply Seaborn style
sns.set()

# Display the chart
plt.show()
```

```python
# Xreff non retrived publishers
data = {'nan DOI': 1937, 'not found': 203, 'https err': 174}
plt.pie(data.values(), labels=data.keys(), autopct='%1.1f%%', startangle=180, colors=sns.color_palette('pastel'))
total = sum(data.values())
plt.title('Xreff non retrived publishers')
plt.legend(title='Total: {}'.format(total))
plt.axis('equal')
sns.set()
plt.show()
```
