```python
import json

def extract_text(data, section_type, output_filename):
    extracted_text = ""
    for passage in data[0]['documents'][0]['passages']:
        if passage['infons']['section_type'] == section_type:
            extracted_text += passage['text'] + " "

    with open(output_filename, 'w', encoding='utf-8', errors='replace') as f:
        f.write(extracted_text)

if __name__ == "__main__":
    with open('17875210.json', 'r') as file: 
        data = json.load(file)

    extract_text(data, 'METHODS', 'METHODS_17875210.txt')
    extract_text(data, 'RESULTS', 'RESULTS_17875210.txt')
    extract_text(data, 'CONCL', 'CONCL_17875210.txt')
    extract_text(data, 'REF', 'REF_17875210.txt')
    extract_text(data, 'FIG', 'FIG_17875210.txt')
    extract_text(data, 'SUPPL', 'SUPPL_17875210.txt')
    extract_text(data, 'DISCUSS', 'DISCUSS_17875210.txt')
```
