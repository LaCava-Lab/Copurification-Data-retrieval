```python
import json

ranges = [(10, 26), (34, 69), (72, 104)]
output_file = 'pmid_17875210'

def extract_text(data, ranges, output_file): 
    extracted_text = ""
    for r in ranges:
        i = r[0]
        while i < r[1]:  
            passage_text = data[0]['documents'][0]['passages'][i]['text']
            extracted_text += passage_text + " "  
            i += 1  # increment i for the next iteration

    with open(output_file + '.txt', 'w', encoding='utf-8', errors='replace') as f:
        f.write(extracted_text)

if __name__ == "__main__": 
    extract_text(data, ranges, output_file)
```
