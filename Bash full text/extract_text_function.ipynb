```python
import json

ranges = [(10, 26), (34, 69), (72, 104)]
output_file = 'pmid_17875210.json'


def extract_text(data, ranges, output_file): # Extracts text from passages and appends it to a JSON file.

    extracted_text = ""
    for r in ranges:
        for i in range(r[0], r[1]):  # extract text from each range
            passage_text = data[0]['documents'][0]['passages'][i]['text']
            extracted_text += passage_text + " "  # concatenate with a space

    with open(output_file, 'w') as f:
        json.dump(extracted_text, f, indent=4)
```

```python
if __name__ == "__main__": 
    extract_text(data, ranges, output_file)
```
