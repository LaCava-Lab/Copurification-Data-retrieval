```python
with open('pmids_quotes_query2024-09-18.txt', 'r') as f1, open('CO_IP_pmids_2024-09-16.txt', 'r') as f2:
    qoutes = set(line.strip() for line in f1)
    old = set(line.strip() for line in f2)

common_pmids = qoutes & old

print("pmids with qoutes:", len(qoutes))
print("common pmids between old and qoutes:", len(common_pmids))
```
